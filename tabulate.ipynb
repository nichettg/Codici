````python
from tabulate import tabulate
from IPython.display import display, Markdown, HTML

# Dati di esempio
data = [[1, 'Alice', 24], [2, 'Bob', 30], [3, 'Charlie', 22]]
headers = ['ID', 'Name', 'Age']

# Genera tabelle
markdown_table = tabulate(data, headers=headers, tablefmt='pipe')
html_table = tabulate(data, headers=headers, tablefmt='html')
latex_table = tabulate(data, headers=headers, tablefmt='latex')

# Mostra Markdown
display(Markdown("### Tabella in formato **Markdown**"))
display(Markdown(markdown_table))

# Mostra HTML
display(Markdown("### Tabella in formato **HTML**"))
display(HTML(html_table))

# Mostra LaTeX come codice
display(Markdown("### Tabella in formato **LaTeX (codice)**"))
display(Markdown(f"```latex\n{latex_table}\n```"))
````

````python
from tabulate import tabulate
from IPython.display import display, HTML

data1 = [[1, 'Alice', 24], [2, 'Bob', 30]]
data2 = [[3, 'Charlie', 22], [4, 'Diana', 27]]
headers = ['ID', 'Name', 'Age']

table1 = tabulate(data1, headers=headers, tablefmt='html')
table2 = tabulate(data2, headers=headers, tablefmt='html')

html = f"""
<div style="display: flex; gap: 40px;">
  <div>{table1}</div>
  <div>{table2}</div>
</div>
"""

display(HTML(html))
````
