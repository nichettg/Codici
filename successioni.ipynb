```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display
```

#### Modifica il modello in base alla funzione da graficare

```python
def modello(x,n):
    return (n * np.exp(-x) + x**2) / (n * np.abs(x))
```

### Funzione limite, convergenza puntuale$$\lim_{n \to +\infty} f_n(x) = f(x)$$ Non viene calcolata automaticamente, serve eseguire il limite e modificare la funzione. 

```python
def limite(x):
    return np.exp(-x)
```

```python
def plot_successione(x_range, start, stop, step, checkbox_value):
    if step <= 0 or start > stop:
        print("⚠️ Assicurati che step > 0 e start ≤ stop.")
        return

    x = np.linspace(-x_range, x_range, 1000)

    plt.figure(figsize=(10, 6))

    # Se la casella è selezionata, disegna la funzione limite
    if checkbox_value:
        f_lim = limite(x)
        plt.plot(x, f_lim, 'k--', label="Limite", linewidth=2)

    for n in range(start, stop + 1, step):
        f_n = modello(x, n)
        plt.plot(x, f_n, label=f"n = {n}")

    plt.title("Successione di funzioni fₙ(x)")
    plt.xlabel("x")
    plt.ylabel("fₙ(x)")
    plt.grid(True)
    plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
    plt.tight_layout()
    plt.show()

slider_x_range = widgets.FloatSlider(
    value=2*np.pi, min=1, max=10*np.pi, step=0.1, description='x_range'
)
slider_start = widgets.IntSlider(value=1, min=1, max=100, step=1, description='start')
slider_stop = widgets.IntSlider(value=50, min=1, max=100, step=1, description='stop')
slider_step = widgets.IntSlider(value=10, min=1, max=50, step=1, description='step')

checkbox = widgets.Checkbox(
    value=False,
    description=f'Mostra funzione limite',
    disabled=False
)

ui = widgets.VBox([
    widgets.HBox([slider_x_range, slider_step]),
    widgets.HBox([slider_start, slider_stop]),
    checkbox
])

interactive_plot = widgets.interactive_output(
    plot_successione,
    {
        'x_range': slider_x_range,
        'start': slider_start,
        'stop': slider_stop,
        'step': slider_step,
        'checkbox_value': checkbox
    }
)

display(ui, interactive_plot)
```
