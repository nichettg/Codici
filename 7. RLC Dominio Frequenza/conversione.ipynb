```python
incertezza_sfasamento = 50
percentuale_ampiezza = 50/100
percentuale_ampiezza_fondoscala = 50/100
```

# Lettura

```python
if grezzi:
    file_input= "data/dati_grezzi copy.csv"
    data = let.csv_to_dict(file_input)
    gf.display_dizionario(data)

if simulati:
    file_input= "data/dati_simulati.csv"
    data = let.csv_to_dict(file_input, tab=True)
    gf.display_dizionario(data)
```

# Conversione

Conversione per $V$.

```python
if grezzi:
    elab_A = {
        'x' : data['f(kHz)'] * 10**3,
        'sx' : np.zeros_like(data['f(kHz)']),
        'y' : data['V(V)'],
        'sy' : percentuale_ampiezza_fondoscala * 0.5 + percentuale_ampiezza * data['V(V)']
    }
    gf.display_dizionario(elab_A)

if simulati:
    A = data['Vout(V)'] / data['Vin(V)']
    sVout = (1/10 * data['f.s.(V)'] + 3/100 * data['Vout(V)'])
    sVin = (1/10 * data['f.s.(V)'] + 3/100 * data['Vin(V)'])
    sA = A * np.sqrt( (sVout / data['Vout(V)'])**2 + (sVin / data['Vin(V)'])**2 )
    elab_A = {
        'x' : data['f(Hz)'],
        'sx' : np.zeros_like(data['f(Hz)']),
        'y' : A,
        'sy' : sA
    }
    del A, sA
    gf.display_dizionario(elab_A)
```

Conversione per $\Delta \Phi$.

```python
if grezzi:
    """elab_phi = {
        'x' : data['f(kHz)'][4:67] * 10**3,
        'sx' : np.zeros_like(data['f(kHz)'][4:67]),
        'y' : data['phi(°)'][4:67],
        'sy' : np.zeros_like(data['phi(°)'][4:67]) + 2
    }"""
    elab_phi = {
        'x' : data['f(kHz)'][4:67] * 10**3,
        'sx' : np.zeros_like(data['f(kHz)'][4:67]),
        'y' : data['phi(°)'][4:67],
        'sy' : np.zeros_like(data['phi(°)'][4:67]) + incertezza_sfasamento
    }
    gf.display_dizionario(elab_phi)

if simulati:
    phi = 2 * np.pi * data['f(Hz)'] * data['dT(us)'] * 10**(-6)
    sphi = 2 * np.pi * data['f(Hz)'] * (1/100 * data['f.s.(us)']) * 10**(-6)
    #sphi = 2 * np.pi * data['f(Hz)'] * (3/100 * data['dT(us)']) * 10**(-6)
    elab_phi = {
        'x' : data['f(Hz)'],
        'sx' : np.zeros_like(data['f(Hz)']),
        'y' : phi,
        'sy' : sphi
    }
    del phi, sphi
    gf.display_dizionario(elab_phi)
```

# Esportazione
