# Conversione

Conversione per $V$.

```python
elab_V = {
    'x' : data['f(kHz)'] * 10**3,
    'sx' : np.zeros_like(data['f(kHz)']),
    'y' : data['V(V)'],
    'sy' : 10/100 * 0.5 + 3/100 * data['V(V)']
}
gf.display_dizionario(elab_V)
```

Conversione per $\Delta \Phi$.

```python
elab_phi = {
    'x' : data['f(kHz)'] * 10**3,
    'sx' : np.zeros_like(data['f(kHz)']),
    'y' : data['phi(°)'],
    'sy' : np.zeros_like(data['phi(°)']) + 2
}
gf.display_dizionario(elab_phi)
```
