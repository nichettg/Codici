# Lettura

```python
file_input= "data/dati_grezzi.csv"
data = let.csv_to_dict(file_input, tab=True)
gf.display_dizionario(data)
```

# Conversione

Conversione per $A$.

```python
A = data['Vout(V)'] / data['Vin(V)']
sVout = (1/10 * data['f.s.(V)'] + 3/100 * data['Vout(V)'])
sVin = (1/10 * data['f.s.(V)'] + 3/100 * data['Vin(V)'])
sA = A * np.sqrt( (sVout / data['Vout(V)'])**2 + (sVin / data['Vin(V)'])**2 )
elab_A = {
    'x' : data['f(Hz)'],
    'sx' : np.zeros_like(data['f(Hz)']),
    'y' : A,
    'sy' : sA
}
del A, sA
gf.display_dizionario(elab_A)
```

Conversione per $\Delta \Phi$.

```python
phi = 2 * np.pi * data['f(Hz)'] * data['dT(μs)'] * 10**(-6)
sphi = 2 * np.pi * data['f(Hz)'] * (1/100 * data['f.s.(μs)']) * 10**(-6)
#sphi = 2 * np.pi * data['f(Hz)'] * (3/100 * data['dT(μs)']) * 10**(-6)
elab_phi = {
    'x' : data['f(Hz)'],
    'sx' : np.zeros_like(data['f(Hz)']),
    'y' : phi,
    'sy' : sphi
}
del phi, sphi
gf.display_dizionario(elab_phi)
```

Conversione per Bode.

```python
A = data['Vout(V)'] / data['Vin(V)']
sVout = (1/10 * data['f.s.(V)'] + 3/100 * data['Vout(V)'])
sVin = (1/10 * data['f.s.(V)'] + 3/100 * data['Vin(V)'])
sA = A * np.sqrt( (sVout / data['Vout(V)'])**2 + (sVin / data['Vin(V)'])**2 )
B = 20 * np.log10(A)
sB = 20 * (sA / (A * np.log(10)))
elab_B = {
    'x' : np.log10(data['f(Hz)']),
    'sx' : np.zeros_like(data['f(Hz)']),
    'y' : B,
    'sy' : sB
}
del A, sA, B, sB
gf.display_dizionario(elab_B)
```

# Esportazione

```python
file_output1 = "data/dati_A.csv"
file_output2 = "data/dati_phi.csv"
file_output3 = "data/dati_bode.csv"

let.dict_to_csv(elab_A, file_output1)
let.dict_to_csv(elab_phi, file_output2)
let.dict_to_csv(elab_B, file_output3)
```
