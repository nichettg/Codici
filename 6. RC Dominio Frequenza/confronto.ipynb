# Confronto

```python
dizionario_f_taglio = {}

# Locale
key = "locale"
tipo = "A"
c = dizionario_ris_fit[key][tipo]["parametri"][0]
sigma_c = dizionario_ris_fit[key][tipo]["s_parametri"][0]
m = dizionario_ris_fit[key][tipo]["parametri"][1]
sigma_m = dizionario_ris_fit[key][tipo]["s_parametri"][1]
dizionario_f_taglio[f"{key}_{tipo}"] = Misura(
    ( 1/np.sqrt(2) - c) / m,
    np.sqrt((sigma_c / m)**2 + ((1/np.sqrt(2) - c) / (m**2) * sigma_m)**2)
)
tipo = "phi"
c = dizionario_ris_fit[key][tipo]["parametri"][0]
sigma_c = dizionario_ris_fit[key][tipo]["s_parametri"][0]
m = dizionario_ris_fit[key][tipo]["parametri"][1]
sigma_m = dizionario_ris_fit[key][tipo]["s_parametri"][1]
dizionario_f_taglio[f"{key}_{tipo}"] = Misura(
    ( 1/2 - c) / m,
    np.sqrt((sigma_c / m)**2 + ((1/2 - c) / (m**2) * sigma_m)**2)
)

# Non Lineare
key = "nonlineare"
for tipo in ["A","phi"]:
    dizionario_f_taglio[f"{key}_{tipo}"] = Misura(
        dizionario_ris_fit[key][tipo]["parametri"][0],
        dizionario_ris_fit[key][tipo]["s_parametri"][0],
    )

# Bode
dizionario_f_taglio["bode"] = Misura(
    dizionario_ris_fit["bode"]["parametri"][0],
    dizionario_ris_fit["bode"]["s_parametri"][0]
)

# Lineare
key = "lineare"
tipo = "A"
c = dizionario_ris_fit[key][tipo]["parametri"][0]
sigma_c = dizionario_ris_fit[key][tipo]["s_parametri"][0]
m = dizionario_ris_fit[key][tipo]["parametri"][1]
sigma_m = dizionario_ris_fit[key][tipo]["s_parametri"][1]
dizionario_f_taglio[f"{key}_{tipo}"] = Misura(
    1 / m,
    sigma_m / m**2
)
tipo = "phi"
c = dizionario_ris_fit[key][tipo]["parametri"][0]
sigma_c = dizionario_ris_fit[key][tipo]["s_parametri"][0]
m = dizionario_ris_fit[key][tipo]["parametri"][1]
sigma_m = dizionario_ris_fit[key][tipo]["s_parametri"][1]
dizionario_f_taglio[f"{key}_{tipo}"] = Misura(
    -1/m,
    sigma_m / m**2
)
```
