```python
import numpy as np
from phaseportrait import PhasePortrait2D

def system(x, v):
    return v,  x - x**3
def U(x):
    return  1/4 * x**4 - 1/2 * x**2

import matplotlib.pyplot as plt

# dominio
x = np.linspace(-2, 2, 400)

# plot
plt.plot(x, U(x))
plt.xlabel("x")
plt.ylabel("U(x)")
plt.title("Grafico dell'energia potenziale")
plt.grid()

plt.show()

pp = PhasePortrait2D(system, [-2, 2])
pp.plot()
```
