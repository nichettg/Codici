```python
# Valori standard
incertezza_sfasamento = 2
percentuale_ampiezza = 3/100
percentuale_ampiezza_fondoscala = 10/100
```

# Lettura

```python
file_input= "data/dati_grezzi_Vin.csv"
data = let.csv_to_dict(file_input, tab=False)
gf.display_dizionario(data)
```

# Conversione

Conversione per $A$.

```python
inizio = 6
fine = 63

sVout = percentuale_ampiezza_fondoscala * 0.5 + percentuale_ampiezza * data['Vout(V)']

sVin = percentuale_ampiezza_fondoscala * 0.5 + percentuale_ampiezza * data['Vin(V)']
#sVin = data['sVin(V)']
#sVin = data['sVin_MC_gauss(V)']
#sVin = data['sVin_MC_unif(V)']

A = data['Vout(V)'] / data['Vin(V)']
sA = A * np.sqrt(
    (sVout / data['Vout(V)'])**2 +
    (sVin / data['Vin(V)'])**2
)

elab_A = {
    'x' : data['f(kHz)'] * 10**3,
    'sx' : np.zeros_like(data['f(kHz)']),
    'y' : A,
    'sy' : sA
}
gf.display_dizionario(elab_A)
```

Conversione per $\Delta \Phi$.

```python
elab_phi = {
    'x' : data['f(kHz)'][inizio:fine] * 10**3,
    'sx' : np.zeros_like(data['f(kHz)'][inizio:fine]),
    'y' : data['phi(°)'][inizio:fine] * (np.pi / 180),
    'sy' : (np.zeros_like(data['phi(°)'][inizio:fine]) + incertezza_sfasamento) * (np.pi / 180),
}
gf.display_dizionario(elab_phi)
```

# Esportazione

```python
file_output1 = "data/dati_ampiezza.csv"
file_output2 = "data/dati_sfasamento.csv"

let.inizializza_output(file_output1)
let.inizializza_output(file_output2)
let.dict_to_csv(elab_A, file_output1)
let.dict_to_csv(elab_phi, file_output2)
```
