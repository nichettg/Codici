### 3.1 Calcolo Area Segnale

##### 3.1.1 Calcolo $N_{ON}$

```python
Non,sNon=[],[]
for frequenze in frequenze_storage:
    Non.append(sum(frequenza for frequenza in frequenze))
    sNon.append(np.sqrt(sum(frequenza for frequenza in frequenze)))
print(f"Somma totale: {Non}\nIncertezza: {sNon}")
```

##### 3.1.2 Calcolo $N_{OFF}$

```python
Noff,sNoff=[],[]
yrdi_gauss = []
for centri,frequenze,a,b in zip(centri_storage,frequenze_storage,a_storage,b_storage):
    somma = 0
    yrdi=[]
    for x,frequenza in zip(centri,frequenze):
        y = a + b*x
        if frequenza > y :
            somma+=y
            yrdi.append(frequenza-y)
        elif frequenza < y :
            somma+=frequenza
            yrdi.append(0)
    Noff.append(somma)
    sNoff.append(np.sqrt(somma))
    yrdi_gauss.append(yrdi.copy())
print(f"Somma inferiore: {Noff}\nIncertezza: {sNoff}")
```

##### 3.1.3 Calcolo $N_{EXC}$

```python
Nexc,sNexc=[],[]
for on,off,son,soff in zip(Non,Noff,sNon,sNoff):
    Nexc.append(on-off)
    sNexc.append(np.sqrt(son**2 + soff**2))
print(f"Somma superiore: {Nexc}\nIncertezza: {sNexc}")
```

##### 3.1.4 Grafico Gaussiane

```python
print(f"xrdi_gauss: {xrdi_gauss}\nyrdi_gauss: {yrdi_gauss}")
```

```python
fig,ax=plt.subplots()
ordine = 1
for titolo, x, y in zip(titoli, xrdi_gauss, yrdi_gauss):
    titolo = titolo.replace("M_","") # Per rimuovere le M dai titoli
    if (titolo == "20min" or titolo == "2h" or titolo == "6h20m"):
        ordine = ordine-1
        gf.parametri_grafico(ax,titolo="Confronto",xlabel="Canali",ylabel="Conteggi in Eccesso")
        ax.bar(x,y,width=xdata[1]-xdata[0],color = next(color_cycle), edgecolor='black', linewidth=0.1, label=titolo, zorder=ordine)
gf.salva_grafico(fig,"./Images/confronto_gauss.pdf",formato="pdf")
```

### 3.2 Compatibilità Segnale con Zero

##### 3.2.1 Test Significanza

```python
a1_storage=[]
for S,sS in zip(Nexc,sNexc):
    comp = S/sS
    a1_storage.append(comp)
print(f"Significanza: {a1_storage}")
```

##### 3.2.2 Test Chi Quadro

```python
b1_storage=[]
for S,sS in zip(Nexc,sNexc):
    comp = (S/sS)**2
    p = chi2.sf(comp, 1)
    z_score = norm.isf(p/2)
    b1_storage.append(z_score)
print(f"z-score: {b1_storage}")
```

### 3.3 Andamento Temporale Significanze

```python
rows=[]
for titolo,c,chi,z in zip(titoli,a1_storage,chi_storage,b1_storage):
    rows.append([titolo,chi,c,z])
headers = ["Serie", "p-value fit", "Significanza", "z-score"]

print(f"Indici:\n{i1}\t{i2}\t{i3}\t{i4}\n")
print(tabulate.tabulate(rows, headers=headers, floatfmt=".5f"))
```
